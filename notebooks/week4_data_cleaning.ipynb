```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd().parent
file_path = BASE_DIR / "data" / "Daily Household Transactions.csv"

df = pd.read_csv(file_path)

print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("First 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("Rows with missing Subcategory or Note:")
display(df[df["Subcategory"].isna() | df["Note"].isna()].head(10))

print("\nDuplicate rows:")
display(df[df.duplicated(keep=False)].sort_values(list(df.columns)).head(20))
```

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
print("Current number of rows:", df.shape[0])
print("Current number of columns:", df.shape[1])
```

```python
df["Subcategory"] = df["Subcategory"].fillna("Unknown")
df["Note"] = df["Note"].fillna("Unknown")

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce")

print("Date data type:", df["Date"].dtype)
print("Invalid dates:", df["Date"].isna().sum())
print("\nDate range:")
print("Start:", df["Date"].min())
print("End:", df["Date"].max())
```

```python
invalid_dates = df[df["Date"].isna()]

print("Number of invalid dates:", len(invalid_dates))
print("\nInvalid date rows:")
display(invalid_dates.head(20))
```

```python
X = df["Note"]
y = df["Category"]

print("Number of transactions:", len(df))
print("Number of categories:", y.nunique())
print("\nTransactions per category:")
print(y.value_counts())
```

```python
category_counts = y.value_counts()

rare_categories = category_counts[category_counts < 5]

print("Number of rare categories:", len(rare_categories))
print("\nRare categories:")
print(rare_categories)
```

```python
category_counts = y.value_counts()

y_model = y.where(y.map(category_counts) >= 5, "Other")

print("Original categories:", y.nunique())
print("ML categories:", y_model.nunique())
print("\nML category distribution:")
print(y_model.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_model,
    test_size=0.2,
    random_state=42,
    stratify=y_model
)

print("Training transactions:", len(X_train))
print("Testing transactions:", len(X_test))
print("Training categories:", y_train.nunique())
print("Testing categories:", y_test.nunique())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
print("Number of features:", len(vectorizer.get_feature_names_out()))
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

print("Model training completed successfully.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
X_combined = (
    df["Note"].astype(str) + " " +
    df["Subcategory"].astype(str) + " " +
    df["Mode"].astype(str)
)

X_train, X_test, y_train, y_test = train_test_split(
    X_combined,
    y_model,
    test_size=0.2,
    random_state=42,
    stratify=y_model
)

print("Training transactions:", len(X_train))
print("Testing transactions:", len(X_test))
print("Training categories:", y_train.nunique())
print("Testing categories:", y_test.nunique())
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
print("Number of features:", len(vectorizer.get_feature_names_out()))
```

```python
model = MultinomialNB()
model.fit(X_train_tfidf, y_train)

print("Improved model training completed successfully.")
```

```python
y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

print("Improved Model Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC()
svm_model.fit(X_train_tfidf, y_train)

print("Linear SVM model training completed successfully.")
```

```python
svm_pred = svm_model.predict(X_test_tfidf)

svm_accuracy = accuracy_score(y_test, svm_pred)

print("Linear SVM Accuracy:", svm_accuracy)
print("\nClassification Report:")
print(classification_report(y_test, svm_pred, zero_division=0))
```
