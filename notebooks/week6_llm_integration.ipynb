```python
import pandas as pd
from pathlib import Path
from google import genai

BASE_DIR = Path.cwd().parent

summary = pd.read_csv(BASE_DIR / "data" / "monthly_financial_summary.csv")
categories = pd.read_csv(BASE_DIR / "data" / "category_spending.csv")

total_income = summary["Income"].sum()
total_expense = summary["Amount"].sum()
total_savings = summary["Savings"].sum()

top_categories = categories.head(5)[["Category", "Amount"]].copy()

prompt = f"""
Analyze the following personal finance data and provide 3 practical savings tips.

Total Income: ₹{total_income:.2f}
Total Expenses: ₹{total_expense:.2f}
Total Savings: ₹{total_savings:.2f}

Top 5 Spending Categories:
{top_categories}

Give simple, personalized and actionable advice based only on this data.
"""

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

print(response.text)
```

```python
def generate_savings_tips(total_income, total_expense, total_savings, top_categories):
    prompt = f"""
    Analyze the following personal finance data and provide 3 practical savings tips.

    Total Income: ₹{total_income:.2f}
    Total Expenses: ₹{total_expense:.2f}
    Total Savings: ₹{total_savings:.2f}

    Top Spending Categories:
    {top_categories}

    Give simple, personalized and actionable advice based only on this data.
    """

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text
```

```python
tips = generate_savings_tips(
    total_income,
    total_expense,
    total_savings,
    top_categories
)

print(tips)
```
